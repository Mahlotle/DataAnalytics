MERGE, JOIN, AND CONCATINATE

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\masem\Downloads\LOTR.csv")
```

```python
df1
```

```python
df2 = pd.read_csv(r"C:\Users\masem\Downloads\LOTR_2.csv")
df2
```

USING MERGE

```python
df1.merge(df2)
```

```python
df1.merge(df2, on='FellowshipID', how='inner')
```

```python
df1.merge(df2, on=['FellowshipID', 'FirstName'], how='inner')
```

```python
df1.merge(df2, how='outer')      
```

```python
df1.merge(df2, how='left')   
```

```python
df1.merge(df2, how='right')   
```

```python
df1.merge(df2, how='cross')   
```

JOINS

```python
df1.join(df2, on = 'FellowshipID', how='outer', lsuffix='_left',rsuffix = '_right'  )
```

```python
df3 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), lsuffix='_left', rsuffix='_right')
df3
```

CONCATINATE

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], join='inner')
```

```python
pd.concat([df1,df2], join='outer')
```

```python
pd.concat([df1,df2], join='outer', axis=1)
```
